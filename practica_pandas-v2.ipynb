```python
import pandas as pd

ruta = "https://raw.githubusercontent.com/VMjavier/claseAnalisisDatos/refs/heads/main/veterinaria_colombia.csv"

df = pd.read_csv(ruta)
df.head()
```

```python
df.info()
```

Limpieza de datos

```python
df.columns
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ','_')
```

```python
df.columns
```

```python
df.rename(columns = {"nombre": "nombre_mascota"}, inplace = True)
```

```python
df.columns
```

Eliminar filas o columnas completamente vacias

```python
df.dropna(axis = 0, how = "all", inplace = True) # Fila = 0
df.dropna(axis = 1, how = "all", inplace = True) # Columna = 1
```

```python
df.shape #(filas, columnas)
```

Eliminar filas o columnas a mi eleccion

```python
df["nombre_mascota"]
```

```python
df.drop(columns = ["nombre_mascota"], inplace = True)
```

```python
df.columns
```

Limpiar columnas con numeros y letras mezclados

```python
df.head(3)
```

```python
df["edad"] = df["edad"].astype(str).str.extract(r'(\d+\.?\d*)')
df["peso_kg"] = df["peso_kg"].astype(str).str.extract(r'(\d+\.?\d*)')
df["temperatura_c"] = df["temperatura_c"].astype(str).str.extract(r'(\d+\.?\d*)')
df["frec_cardiaca"] = df["frec_cardiaca"].astype(str).str.extract(r'(\d+\.?\d*)')
df["frec_respiratoria"] = df["frec_respiratoria"].astype(str).str.extract(r'(\d+\.?\d*)')
df["costo_consulta"] = df["costo_consulta"].astype(str).str.extract(r'(\d+\.?\d*)')
```

```python
df.head(3)
```

Convertir variables a int o float

```python
df.info()
```

```python
df["edad"] = pd.to_numeric(df["edad"], errors = "coerce")
df["peso_kg"] = pd.to_numeric(df["peso_kg"], errors = "coerce")
df["temperatura_c"] = pd.to_numeric(df["temperatura_c"], errors = "coerce")
df["frec_cardiaca"] = pd.to_numeric(df["frec_cardiaca"], errors = "coerce")
df["frec_respiratoria"] = pd.to_numeric(df["frec_respiratoria"], errors = "coerce")
df["costo_consulta"] = pd.to_numeric(df["costo_consulta"], errors = "coerce")
df.info()
```

```python
df.head()
```

Quitar duplicados

```python
df.drop_duplicates(inplace = True)
```

```python
df.shape
```

Datos nulos

```python
df.isnull().sum()
```

Rellenar valores faltantes tipo object

```python
df["raza"] = df["raza"].fillna("No registra")
df["especie"] = df["especie"].fillna("Sin especie")
df["id_mascota"] = df["id_mascota"].fillna("Sin ID")
```

```python
df.isnull().sum()
```

Rellenar valores faltantes numericos, no tipo object

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.set_style("whitegrid")  # activa cuadricula
sns.histplot(
    df["edad"],
    kde=True,
    color="#62109F",  # color solicitado
    alpha=0.7
)

plt.title("Distribución de edad", fontsize=16)
plt.xlabel("edad")
plt.ylabel("Frecuencia")
plt.show()
```

```python
import seaborn as sns

sns.boxplot(x=df["edad"], color ="#62109F")
plt.title("Bloxplot edad", fontsize=16)
plt.show()
```

```python
df.isnull().sum()
```

```python
#mean = Promedio = media
#median = mediana
media = df["edad"].mean()
mediana = df["edad"].median()

print(f"Media={media}")
print(f"Mediana={mediana}")

df['edad'] = df['edad'].fillna(mediana)
```

```python
df.isnull().sum()
```

```python
#Completar datos con el promedio o media
mediana = df["edad"].median()
df['edad'] = df['edad'].fillna(mediana)
mediana = df["peso_kg"].median()
df['peso_kg'] = df['peso_kg'].fillna(mediana)
mediana = df["temperatura_c"].median()
df['temperatura_c'] = df['temperatura_c'].fillna(mediana)
mediana = df["frec_cardiaca"].median()
df['frec_cardiaca'] = df['frec_cardiaca'].fillna(mediana)
mediana = df["frec_respiratoria"].median()
df['frec_respiratoria'] = df['frec_respiratoria'].fillna(mediana)
mediana = df["costo_consulta"].median()
df['costo_consulta'] = df['costo_consulta'].fillna(mediana)
```

```python
df.isnull().sum()
```

Exportar el dataset limpio

```python
import pandas as pd

df.to_excel("mascotas_limpio.xlsx",index=False)
df.to_csv("mascotas_limpio.csv", index=False)
```
